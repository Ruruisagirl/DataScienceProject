```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster
from folium.plugins import MousePosition
from folium.features import DivIcon
```

```python
# Load the launch site data from a CSV file
launch_sites_df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv')

# Select the relevant columns for launch sites (Site Name, Latitude, Longitude)
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long', 'class']]

# Remove duplicate launch sites to get unique locations
launch_sites_df = launch_sites_df.groupby(['Launch Site'], as_index=False).first()

# Display the unique launch sites
display(launch_sites_df)
```

```python
# Get the average latitude and longitude to center the map
avg_lat = launch_sites_df['Lat'].mean()
avg_long = launch_sites_df['Long'].mean()

# Create a base map centered around the average location
site_map = folium.Map(location=[avg_lat, avg_long], zoom_start=5)

# Add a tile layer to the map
folium.TileLayer('OpenStreetMap').add_to(site_map)

# Iterate through each launch site and add a marker
for index, site in launch_sites_df.iterrows():
    # Create a marker for the launch site
    marker = folium.Marker(
        location=[site['Lat'], site['Long']],  # Latitude and Longitude of the launch site
        popup=site['Launch Site'],  # Display the launch site name when the marker is clicked
        icon=folium.Icon(color='blue', icon='info-sign')  # Set the marker color and icon
    )
    # Add the marker to the map
    marker.add_to(site_map)

# Display the map
display(site_map)
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Replace these with the coordinates obtained from MousePosition on the map
coastline_point_lat = 28.563197
coastline_point_lon = -80.56768

# Choose a launch site from the launch_sites_df to calculate the distance to
# For example, using the first launch site (CCAFS LC-40)
launch_site_lat = launch_sites_df.loc[0, 'Lat']
launch_site_lon = launch_sites_df.loc[0, 'Long']
launch_site_name = launch_sites_df.loc[0, 'Launch Site']

# Calculate the distance
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_point_lat, coastline_point_lon)

print(f"Distance from {launch_site_name} to coastline: {distance_coastline:.2f} km")

# Optionally, add a line to the map showing the distance
points = [[launch_site_lat, launch_site_lon], [coastline_point_lat, coastline_point_lon]]
folium.PolyLine(points, color='blue', weight=2.5, opacity=0.5).add_to(site_map)

# Add a marker for the coastline point
folium.Marker(
    location=[coastline_point_lat, coastline_point_lon],
    popup=f'Coastline ({distance_coastline:.2f} km)',
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % f"{distance_coastline:.2f} km",
    )
).add_to(site_map)

display(site_map)
```

```python
# Calculate and visualize distances from all launch sites to coastline, city, and highway

proximity_points = {
    'Coastline': coastline_coordinate,
    'City': city_coordinate,
    'Highway': highway_coordinate
}

colors = {
    'Coastline': 'blue',
    'City': 'red',
    'Highway': 'purple'
}

for index, site in launch_sites_df.iterrows():
    launch_site_lat = site['Lat']
    launch_site_lon = site['Long']
    launch_site_name = site['Launch Site']

    for point_name, coords in proximity_points.items():
        point_lat, point_lon = coords
        distance = calculate_distance(launch_site_lat, launch_site_lon, point_lat, point_lon)

        print(f"Distance from {launch_site_name} to {point_name}: {distance:.2f} km")

        # Add a polyline connecting the launch site to the proximity point
        points = [[launch_site_lat, launch_site_lon], [point_lat, point_lon]]
        folium.PolyLine(points, color=colors[point_name], weight=2.5, opacity=0.5).add_to(site_map)

        # Add a marker with the distance on the proximity point
        folium.Marker(
            location=[point_lat, point_lon],
            popup=f'{point_name} ({distance:.2f} km)',
            icon=DivIcon(
                icon_size=(20,20),
                icon_anchor=(0,0),
                html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % f"{distance:.2f} km",
            )
        ).add_to(site_map)

# Display the updated map
display(site_map)
```
